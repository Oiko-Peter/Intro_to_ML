```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Settings for better visualizations
pd.set_option('display.max_columns', None)

# Load the data using relative path
train_data = pd.read_csv('../Data/train.csv')
test_data = pd.read_csv('../Data/test.csv')

# Look at the data shape
print("Train data shape:", train_data.shape)
print("Test data shape:", test_data.shape)

# Look at first few rows
train_data.head()
```

```python
# Check missing values
missing_values = train_data.isnull().sum()
missing_values = missing_values[missing_values > 0].sort_values(ascending=False)
print("\nMissing Values:")
print(missing_values)
```

```python
# Basic statistics of SalePrice
print("\nSalePrice Statistics:")
print(train_data['SalePrice'].describe())

# Plot SalePrice distribution
plt.figure(figsize=(10, 6))
sns.histplot(train_data['SalePrice'], kde=True)
plt.title('SalePrice Distribution')
plt.tight_layout()

# Log transform of SalePrice
plt.figure(figsize=(10, 6))
sns.histplot(np.log1p(train_data['SalePrice']), kde=True)
plt.title('Log Transformed SalePrice Distribution')
plt.tight_layout()
```

```python
# Select numeric columns
numeric_cols = train_data.select_dtypes(include=[np.number]).columns

# Calculate correlations with SalePrice
correlations = train_data[numeric_cols].corr()['SalePrice'].sort_values(ascending=False)
print("\nTop 25 Correlations with SalePrice:")
print(correlations[:15])

# Correlation heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(train_data[correlations[:10].index].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Top Features')
plt.tight_layout()
```

```python
# Separate features and target in training data
X_train = train_data.drop(['SalePrice', 'Id'], axis=1)
y_train = train_data['SalePrice']

# Prepare test data
X_test = test_data.drop('Id', axis=1)

# Since we'll be evaluated on log of predictions, let's transform y
y_train_log = np.log(y_train)

# Basic handling of missing values (we'll improve this later)
# For numeric columns, fill with median
numeric_cols = X_train.select_dtypes(include=[np.number]).columns
X_train[numeric_cols] = X_train[numeric_cols].fillna(X_train[numeric_cols].median())
X_test[numeric_cols] = X_test[numeric_cols].fillna(X_test[numeric_cols].median())

# For categorical columns, fill with 'None'
categorical_cols = X_train.select_dtypes(exclude=[np.number]).columns
X_train[categorical_cols] = X_train[categorical_cols].fillna('None')
X_test[categorical_cols] = X_test[categorical_cols].fillna('None')

# Convert categorical variables to dummy variables
X_train = pd.get_dummies(X_train)
X_test = pd.get_dummies(X_test)

# Make sure train and test have same columns
common_cols = set(X_train.columns) & set(X_test.columns)
X_train = X_train[list(common_cols)]
X_test = X_test[list(common_cols)]
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestRegressor

# Create and train model
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train_log)

# Cross validation to check performance
scores = cross_val_score(rf_model, X_train, y_train_log, 
                        scoring='neg_root_mean_squared_error', 
                        cv=5)

print("Cross-validation RMSE scores:", -scores)
print("Average RMSE:", -scores.mean())

# Make predictions on test set
test_predictions = np.exp(rf_model.predict(X_test))

# Create submission DataFrame
submission = pd.DataFrame({
    'Id': test_data['Id'],
    'SalePrice': test_predictions
})

print("\nShape of submission:", submission.shape)
submission.head()
```

```python
from sklearn.ensemble import GradientBoostingRegressor

# Train Gradient Boosting with engineered features
gb_model = GradientBoostingRegressor(n_estimators=100, 
                                    learning_rate=0.1,
                                    max_depth=4,
                                    random_state=42)
gb_model.fit(X_train, y_train_log)

# Cross validation
scores = cross_val_score(gb_model, X_train, y_train_log, 
                        scoring='neg_root_mean_squared_error', 
                        cv=5)

print("Cross-validation RMSE scores:", -scores)
print("Average RMSE:", -scores.mean())

# Look at new feature importances
importance = pd.DataFrame({
    'feature': X_train.columns,
    'importance': gb_model.feature_importances_
})
importance = importance.sort_values('importance', ascending=False).head(10)

print("\nTop 10 most important features:")
print(importance)

# Make predictions and create submission
test_predictions = np.exp(gb_model.predict(X_test))
submission = pd.DataFrame({
    'Id': test_data['Id'],
    'SalePrice': test_predictions
})

# Save submission
submission.to_csv('gb_submission_engineered.csv', index=False)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV

# Define parameter grid
param_grid = {
    'learning_rate': [0.08, 0.1, 0.12],
    'n_estimators': [175, 200, 225],
    'max_depth': [3]
}

# Create base model
gb_model = GradientBoostingRegressor(random_state=42)

# Perform grid search
grid_search = GridSearchCV(
    estimator=gb_model,
    param_grid=param_grid,
    cv=5,
    scoring='neg_root_mean_squared_error',
    verbose=1
)

# Fit the grid search
grid_search.fit(X_train, y_train_log)

# Print results
print("Best parameters:", grid_search.best_params_)
print("Best RMSE:", -grid_search.best_score_)

# Use best model for predictions
best_model = grid_search.best_estimator_
test_predictions = np.exp(best_model.predict(X_test))

# Create submission with tuned model
submission = pd.DataFrame({
    'Id': test_data['Id'],
    'SalePrice': test_predictions
})

# Save submission
submission.to_csv('gb_tuned_submission.csv', index=False)

```

```python
# Feature importance of best model
importance = pd.DataFrame({
    'feature': X_train.columns,
    'importance': best_model.feature_importances_
})
importance = importance.sort_values('importance', ascending=False).head(10)
print("\nTop 10 most important features with tuned model:")
print(importance)
```

```python
# Save submission file
submission.to_csv('rf_submission.csv', index=False)
```

Feature Engineering

```python
# Calculate percentage of missing values
missing_percentages = (train_data.isnull().sum() / len(train_data)) * 100
missing_percentages = missing_percentages.sort_values(ascending=False)

# Show columns with missing values and their percentages
print("Percentage of missing values by column:")
print(missing_percentages[missing_percentages > 0])

# Drop columns with more than 80% missing values
high_missing_cols = missing_percentages[missing_percentages > 80].index.tolist()
print("\nColumns to be dropped (>80% missing):")
print(high_missing_cols)

# Create new feature sets without these columns
X_train = train_data.drop(['SalePrice', 'Id'] + high_missing_cols, axis=1)
X_test = test_data.drop(['Id'] + high_missing_cols, axis=1)
y_train = train_data['SalePrice']
y_train_log = np.log(y_train)

# Handle missing values
numeric_cols = X_train.select_dtypes(include=[np.number]).columns
X_train[numeric_cols] = X_train[numeric_cols].fillna(X_train[numeric_cols].median())
X_test[numeric_cols] = X_test[numeric_cols].fillna(X_test[numeric_cols].median())

categorical_cols = X_train.select_dtypes(exclude=[np.number]).columns
X_train[categorical_cols] = X_train[categorical_cols].fillna('None')
X_test[categorical_cols] = X_test[categorical_cols].fillna('None')


# Add engineered features here
# Quality and Area interactions
X_train['Qual_Living_Area'] = X_train['OverallQual'] * X_train['GrLivArea']
X_test['Qual_Living_Area'] = X_test['OverallQual'] * X_test['GrLivArea']

# Total Square Footage
X_train['TotalSF'] = X_train['GrLivArea'] + X_train['TotalBsmtSF']
X_test['TotalSF'] = X_test['GrLivArea'] + X_test['TotalBsmtSF']

# Garage Quality Interaction
X_train['Garage_Quality'] = X_train['OverallQual'] * X_train['GarageCars']
X_test['Garage_Quality'] = X_test['OverallQual'] * X_test['GarageCars']


# Convert categorical variables
X_train = pd.get_dummies(X_train)
X_test = pd.get_dummies(X_test)

# Ensure same columns
common_cols = set(X_train.columns) & set(X_test.columns)
X_train = X_train[list(common_cols)]
X_test = X_test[list(common_cols)]
```

```python
from xgboost import XGBRegressor

# Create model with same parameters
xgb_model = XGBRegressor(
    learning_rate=0.1,
    max_depth=3,
    n_estimators=200,
    random_state=42
)

# Cross validation
scores = cross_val_score(xgb_model, X_train, y_train_log, 
                        scoring='neg_root_mean_squared_error', 
                        cv=5)

print("\nXGBoost RMSE scores:", -scores)
print("Average RMSE:", -scores.mean())

# Train final model and get feature importance
xgb_model.fit(X_train, y_train_log)

# Feature importance
importance = pd.DataFrame({
    'feature': X_train.columns,
    'importance': xgb_model.feature_importances_
})
importance = importance.sort_values('importance', ascending=False).head(10)
print("\nTop 10 most important features:")
print(importance)

# Make predictions and create submission
test_predictions = np.exp(xgb_model.predict(X_test))
submission = pd.DataFrame({
    'Id': test_data['Id'],
    'SalePrice': test_predictions
})
submission.to_csv('xgb_with_features_submission.csv', index=False)
```
